```python
import os
import torch
import torchaudio
import pandas as pd
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchaudio.transforms import MelSpectrogram, Resample
import sentencepiece as spm

# Dataset Class
class UrduSpeechDataset(Dataset):
    def __init__(self, dataframe, audio_dir, tokenizer, sample_rate=16000):
        self.data = dataframe
        self.audio_dir = audio_dir
        self.tokenizer = tokenizer
        self.sample_rate = sample_rate
        self.mel_transform = MelSpectrogram(sample_rate=sample_rate, n_fft=400, hop_length=160, n_mels=80)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        audio_file = self.data.iloc[idx]["path"]
        sentence = self.data.iloc[idx]["sentence"]

        audio_path = os.path.join(self.audio_dir, audio_file)
        waveform, sr = torchaudio.load(audio_path)

        if sr != self.sample_rate:
            resampler = Resample(orig_freq=sr, new_freq=self.sample_rate)
            waveform = resampler(waveform)

        mel_spec = self.mel_transform(waveform).squeeze(0)

        transcript = torch.tensor(self.tokenizer.encode(sentence), dtype=torch.long)

        return mel_spec, transcript


def collate_fn(batch):
    audio, transcripts = zip(*batch)

    audio_lens = [a.shape[1] for a in audio]
    padded_audio = torch.zeros(len(audio), max(audio_lens), audio[0].shape[0])
    for i, a in enumerate(audio):
        padded_audio[i, :a.shape[1]] = a.T

    transcript_lens = [len(t) for t in transcripts]
    padded_transcripts = torch.zeros(len(transcripts), max(transcript_lens), dtype=torch.long)
    for i, t in enumerate(transcripts):
        padded_transcripts[i, :len(t)] = t

    return padded_audio, audio_lens, padded_transcripts, transcript_lens


class ConformerBlock(nn.Module):
    def __init__(self, dim, num_heads, ff_dim):
        super(ConformerBlock, self).__init__()
        self.ff1 = nn.Linear(dim, ff_dim)
        self.ff2 = nn.Linear(ff_dim, dim)
        self.attention = nn.MultiheadAttention(embed_dim=dim, num_heads=num_heads)
        self.layer_norm1 = nn.LayerNorm(dim)
        self.layer_norm2 = nn.LayerNorm(dim)
        self.conv = nn.Conv1d(dim, dim, kernel_size=15, padding=7, groups=dim)
        self.layer_norm3 = nn.LayerNorm(dim)

    def forward(self, x):
        # Feed-forward
        ff_out = self.ff2(torch.relu(self.ff1(x)))
        x = self.layer_norm1(x + ff_out)

        # Multi-head attention
        attn_out, _ = self.attention(x, x, x)
        x = self.layer_norm2(x + attn_out)

        # Convolution
        conv_out = self.conv(x.permute(0, 2, 1)).permute(0, 2, 1)
        x = self.layer_norm3(x + conv_out)

        return x


# Mixture of Experts Layer
class MixtureOfExperts(nn.Module):
    def __init__(self, input_dim, num_experts, top_k):
        super(MixtureOfExperts, self).__init__()
        assert top_k <= num_experts, "top_k must be less than or equal to num_experts"
        self.experts = nn.ModuleList([nn.Linear(input_dim, input_dim) for _ in range(num_experts)])
        self.gating_network = nn.Linear(input_dim, num_experts)
        self.top_k = top_k

    def forward(self, x):
        gate_scores = torch.softmax(self.gating_network(x), dim=-1)

        top_k_values, top_k_indices = torch.topk(gate_scores, self.top_k, dim=-1)

        expert_outputs = torch.stack([expert(x) for expert in self.experts], dim=-1)

        # Gather top-k experts for each input
        top_k_expert_outputs = torch.gather(
            expert_outputs,
            -1,
            top_k_indices.unsqueeze(-2).expand(-1, -1, x.size(-1), -1),
        )

        weighted_outputs = top_k_expert_outputs * top_k_values.unsqueeze(-2)

        output = weighted_outputs.sum(dim=-1)

        return output



# Conformer-MoE Model
class ConformerMoE(nn.Module):
    def __init__(self, input_dim, num_blocks, num_heads, ff_dim, num_experts, top_k, vocab_size):
        super(ConformerMoE, self).__init__()
        self.blocks = nn.ModuleList([ConformerBlock(input_dim, num_heads, ff_dim) for _ in range(num_blocks)])
        self.moe = MixtureOfExperts(input_dim, num_experts, top_k)
        self.ctc_layer = nn.Linear(input_dim, vocab_size)

    def forward(self, x):
        for block in self.blocks:
            x = block(x)
        x = self.moe(x)
        x = self.ctc_layer(x)
        return x


# Training Function
def train_model(model, train_loader, tokenizer, num_epochs=10, learning_rate=1e-4):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
    criterion = nn.CTCLoss(blank=0, zero_infinity=True)

    for epoch in range(num_epochs):
        model.train()
        epoch_loss = 0
        for batch in train_loader:
            audio, audio_lens, transcripts, transcript_lens = batch

            # Move data to GPU
            audio = audio.to(device)
            transcripts = transcripts.to(device)
            audio_lens = torch.tensor(audio_lens, dtype=torch.long).to(device)
            transcript_lens = torch.tensor(transcript_lens, dtype=torch.long).to(device)

            optimizer.zero_grad()

            # Forward pass
            logits = model(audio)
            logits = logits.log_softmax(2)

            # Compute CTC loss
            loss = criterion(
                logits.permute(1, 0, 2),
                transcripts,
                input_lengths=audio_lens,
                target_lengths=transcript_lens,
            )
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
        print(f"Epoch {epoch + 1}, Loss: {epoch_loss / len(train_loader)}")


# Load Dataset and Train
tokenizer = spm.SentencePieceProcessor(model_file='urdu_tokenizer.model')

df_train = pd.read_csv('/content/df_train.csv')  # Replace with actual path
audio_dir = '/content/extracted_dataset/ur/clips'  # Replace with actual path

train_dataset = UrduSpeechDataset(df_train, audio_dir, tokenizer)
train_loader = DataLoader(train_dataset, batch_size=8, collate_fn=collate_fn)

model = ConformerMoE(input_dim=80, num_blocks=4, num_heads=4, ff_dim=512, num_experts=2, top_k=1, vocab_size=len(tokenizer))

train_model(model, train_loader, tokenizer)
```

```python
from transformers import WhisperForConditionalGeneration, WhisperProcessor
import torch
from torch.utils.data import DataLoader
import pandas as pd
import torchaudio
from torch.utils.data import Dataset
import torch.nn.functional as F
import os
from torch.cuda.amp import autocast, GradScaler

class UrduSpeechDataset(Dataset):
    def __init__(self, dataframe, audio_dir, tokenizer, sample_rate=16000):
        self.data = dataframe
        self.audio_dir = audio_dir
        self.tokenizer = tokenizer
        self.sample_rate = sample_rate

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        audio_file = self.data.iloc[idx]["path"]
        sentence = self.data.iloc[idx]["sentence"]
        audio_path = os.path.join(self.audio_dir, audio_file)
        waveform, sr = torchaudio.load(audio_path)
        if sr != self.sample_rate:
            resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=self.sample_rate)
            waveform = resampler(waveform)
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0, keepdim=True)
        transcript = self.tokenizer(sentence, return_tensors="pt").input_ids.squeeze(0)
        waveform = (waveform - waveform.mean()) / waveform.std()
        return waveform.squeeze(0), transcript

def collate_fn(batch):
    max_length = max([waveform.size(0) for waveform, _ in batch])
    padded_waveforms = []
    transcripts = []
    for waveform, transcript in batch:
        padded_waveform = F.pad(waveform, (0, max_length - waveform.size(0)), value=0)
        padded_waveforms.append(padded_waveform)
        transcripts.append(transcript)
    return torch.stack(padded_waveforms), transcripts

model_name = "openai/whisper-tiny"
processor = WhisperProcessor.from_pretrained(model_name)
model = WhisperForConditionalGeneration.from_pretrained(model_name)
model.config.use_gradient_checkpointing = True

tokenizer = processor.tokenizer
df_train = pd.read_csv('/content/df_train.csv')
audio_dir = '/content/extracted_dataset/ur/clips'

train_dataset = UrduSpeechDataset(df_train, audio_dir, tokenizer)
train_loader = DataLoader(train_dataset, batch_size=2, shuffle=True, collate_fn=collate_fn)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.cuda.empty_cache()
model.to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)
scaler = GradScaler()

def train_whisper_model(model, train_loader, optimizer, device, num_epochs=10, accumulation_steps=4):
    model.train()
    loss_fn = torch.nn.CrossEntropyLoss()
    for epoch in range(num_epochs):
        epoch_loss = 0
        optimizer.zero_grad()
        for step, (waveform, transcripts) in enumerate(train_loader):
            waveform = waveform.to(device)
            transcripts = [transcript.to(device) for transcript in transcripts]
            waveform_cpu = waveform.cpu()
            waveform_cpu = waveform_cpu.unsqueeze(1)
            inputs = processor(waveform_cpu, return_tensors="pt", sampling_rate=16000, padding=True)
            inputs = {key: value.to(device) for key, value in inputs.items()}
            labels = torch.nn.utils.rnn.pad_sequence(transcripts, batch_first=True, padding_value=0).to(device)
            with autocast():
                outputs = model(input_values=inputs["input_values"], labels=labels)
                loss = outputs.loss
            scaler.scale(loss).backward()
            if (step + 1) % accumulation_steps == 0:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()
            epoch_loss += loss.item()
        print(f"Epoch {epoch + 1}, Loss: {epoch_loss / len(train_loader)}")
        torch.cuda.empty_cache()

train_whisper_model(model, train_loader, optimizer, device)
```
